"The latitude of Vienna, Austria is 48.210033, and the longitude is 16.363449." <br>
Some other tiles available:  'OpenStreetMap', 'Cartodb Positron', 'Cartodb dark_matter'

url for the geo.json file to get the boundary of the districts = "https://raw.githubusercontent.com/ginseng666/GeoJSON-TopoJSON-Austria/master/2021/simplified-99.9/bezirke_999_geo.json". I have downloaded this file in the notebook, and will reference it that way (user can also use th url `Last Accessed: 14/08/2026`).

```python
m = folium.Map(
    location= [48.210033, 16.3634],
    zoom_start= 13,
    tiles= 'Cartodb Positron'
)

m
```

```python
def func(x):
    x = {
        'fillColor': '#000000',
        'color' : '#000000', # border,
        'fillOpacity': 0.5, # color inside border
        'dashArray': '4'
    }
    return x


folium.GeoJson(
    '/workspaces/Vienna-Airbnb-Data-Analysis/data/bezirke_999_geo.json',
    name= 'geojson',
    style_function= func(x)
).add_to(m)

m
```
